```python
#importing the basic library
import numpy as np
from matplotlib import pyplot as plt
```

```python
x=np.arange(1,11)
```

```python
x
```

```python
y=2*x
```

```python
y
```

Line chart

```python
plt.plot(x,y)
plt.title("Line chart")
plt.xlabel("x axis") 

plt.ylabel("Y axis")
plt.show()
```

Bar chart

```python
plt.bar(x,y)
plt.title("Bar chart")
plt.xlabel("x axis")
plt.ylabel("Y axis")
plt.show()
```

scatter Plot

```python
a=(1,5,8,6,3,7,9,4)
b=(10,55,21,64,85,33,44,28)
    
plt.scatter(a,b)
plt.title("scatter Plot")
plt.xlabel("x axis")
plt.ylabel("Y axis")
plt.show()
```

Histogram

```python
H=[1,1,1,2,6,6,6,6,4,4,4,4,4,4,3,3,10,10,10,10,2,2,7,7,7,8,8,8,8,8]
```

```python
plt.hist(H)
plt.show()
```

Box plot

```python
B=[1,2,3,4,5,6,7,8,9]
```
